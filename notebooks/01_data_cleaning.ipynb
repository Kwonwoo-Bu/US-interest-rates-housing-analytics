This notebook cleans the raw interest rate and housing market datasets before combining them for analysis. I standardized column names, converted dates and numeric values to the correct data types, and removed rows with invalid dates where needed.

```python
import pandas as pd
from pathlib import Path
```

```python
RAW_DATA = Path("../data/raw")
RAW_DATA.exists()
```

## 3. Cleaning New Home Sales Data

I selected the monthly new home sales data from the Excel workbook, extracted the relevant date and sales columns, converted them to the correct data types, and checked for missing values.

```python
new_home_sales_monthly = pd.read_excel(
    RAW_DATA/"sold_cust.xlsx",
    sheet_name="Monthly"
)
```

```python
new_home_sales_monthly.head(10)
```

```python
new_home_sales_monthly.shape
```

```python
new_home_sales_clean = new_home_sales_monthly.iloc[5:,[0,6]]
new_home_sales_clean.head()
```

```python
data = pd.to_datetime(
    new_home_sales_clean.iloc[:, 0],
    errors="coerce"
)
```

```python
valid_rows = data.notna()
valid_rows.tail()
```

```python
new_home_sales_clean = new_home_sales_clean.loc[
    valid_rows,
    [new_home_sales_clean.columns[0],
    new_home_sales_clean.columns[1]
    ]
].copy()
```

```python
new_home_sales_clean.columns = ["date", "new_home_sales"]
new_home_sales_clean.tail()
```

```python
new_home_sales_clean.dtypes
```

```python
new_home_sales_clean["date"] = pd.to_datetime(
    new_home_sales_clean["date"],
    errors="coerce"
)
```

```python
new_home_sales_clean.dtypes
```

```python
new_home_sales_clean["date"].isna().sum()
```

```python
new_home_sales_clean.shape
```

```python
new_home_sales_clean["new_home_sales"] = pd.to_numeric(
    new_home_sales_clean["new_home_sales"],
    errors="coerce"
)
```

```python
new_home_sales_clean.dtypes
```

```python
new_home_sales_clean["new_home_sales"].isna().sum()
```

```python
new_home_sales_clean.to_csv(
    "../data/processed/new_home_sales_clean.csv",
    index=False
)
```
